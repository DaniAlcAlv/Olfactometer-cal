```python
import os

directory = r"C:\Data\Rerunning previous sessions\13B-828425_2026-01-28T203947Z\testing_2 2-9-2026\behavior\AnalogInput.harp"

for filename in os.listdir(directory):
    full_path = os.path.join(directory, filename)

    # Skip directories
    if os.path.isdir(full_path):
        continue

    # If file has no extension
    if '.' not in filename:
        new_name = filename + ".bin"
        new_path = os.path.join(directory, new_name)
        os.rename(full_path, new_path)
        print(f"Renamed: {filename} -> {new_name}")
```

```python
import os

directory = r"/path/to/your/folder"
old = "oldpart"
new = "newpart"

for filename in os.listdir(directory):
    full_path = os.path.join(directory, filename)

    # Skip directories
    if os.path.isdir(full_path):
        continue

    # Only rename if the old string is in the filename
    if old in filename:
        new_name = filename.replace(old, new)
        new_path = os.path.join(directory, new_name)
        os.rename(full_path, new_path)
        print(f"Renamed: {filename} -> {new_name}")
```
