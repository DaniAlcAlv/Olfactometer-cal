```python
import pandas as pd
import os
from pathlib import Path

CSV_NAME = "RF.csv"
JSON_NAME = "RF.json"
BASE_DIR = Path(r"C:\git\Olfactometer-cal\data")

csv_path = os.path.join(BASE_DIR,  CSV_NAME)
json_path = os.path.join(BASE_DIR, JSON_NAME)

def convert_csv_to_json() -> None:
    print(csv_path)
    df = pd.read_csv(csv_path)

    # Convert to JSON
    df.to_json(json_path, orient="records", indent=4)

    print(f"Saved {csv_path} to {json_path}")

def convert_json_to_csv() -> None:
    print(json_path)
    df = pd.read_json(json_path)

    # Convert to CSV
    df.to_csv(csv_path, index=False)

    print(f"Saved {json_path} to {csv_path}")

convert_json_to_csv()    
```
